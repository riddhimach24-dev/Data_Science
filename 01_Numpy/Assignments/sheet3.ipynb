## Q1. Create `arr = np.array([10,20,30,40,50,60,70,80])`. Print **20, 50, 80** using fancy indexing.

```python
arr = np.array([10,20,30,40,50,60,70,80])
arr[[1,4,7]]
```

## Q2. Using the same array, print **10, 30, 70** and then **40, 60** using fancy indexing.

```python
print(arr[[0,2,6]])
print(arr[[3,5]])
```

## Q3. Create a **3×3** matrix. Print the **first row** and **third row** using fancy indexing.

```python
matrix = np.arange(1,10).reshape(3,3)
print(matrix)
print(matrix[[0]])
print(matrix[[2]])
```

## Q4. Using the same matrix, print the **first column** and **last column** using fancy indexing.

```python

print(matrix[[0,1,2],[0,0,0]])
print(matrix[[0,1,2],[2,2,2]])
```

## Q5. Create `arr = np.array([5,12,18,7,20,25,30,2])`. Print only elements **greater than 15**.

```python
arr = np.array([5,12,18,7,20,25,30,2])
arr[arr > 15]
```

## Q6. Using the same array, print only the **even numbers**.

```python
print(arr[arr%2 == 0])
```

## Q7. Using the same array, print only the numbers **less than 20**.

```python
print(arr[arr<20])
```

## Q8. Create `arr = np.array([18,5,27,10,3,15])`. Sort the array using `np.sort()`.

```python
arr = np.array([18,5,27,10,3,15])
print(np.sort(arr))
```

## Q9. Using the same array, find the indices that would sort the array using `np.argsort()`.

```python
print(np.argsort(arr))
```

## Q10. Create `arr1 = np.array([1,2,3])` and `arr2 = np.array([4,5,6])`. Join both arrays using `np.concatenate()`.

```python
import numpy as np

arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
np.concatenate((arr1,arr2))

```
