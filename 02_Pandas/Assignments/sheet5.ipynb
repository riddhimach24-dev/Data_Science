# Q40. ⭐ Final Practical Challenge

Using the same 8-column DataFrame:

1. Sort employees by `Salary` from highest to lowest.
2. Find the average Salary of each Department.
3. Find the Department with the highest average Salary.
4. Find the average Score by City.
5. Create a pivot table of mean Salary by Department and City.
6. Create a crosstab of Department vs Gender.

**Expected Result:** A small analysis report containing all six results.

```python
# print("Original DataFrame")
# print(A)

A = A.sort_values(by= 'Salary', ascending= False)
print(A)

avg_salary = A.groupby('Department')['Salary'].mean()
print("\nAverage salary of each department:")
print(avg_salary)

highest_avg_salary= A.groupby('Department')['Salary'].mean().idxmax()
print("\nDepartment with highest average salary:")
print(highest_avg_salary)

avg_score = A.groupby('City')['Score'].mean()
print("\nAverage score by city:")
print(avg_score)

pivot_salary = A.pivot_table(values= 'Salary', index= 'Department', columns='City', aggfunc= 'mean')
print("\nPivot table: Average Salary by Department and City")
print(pivot_salary)

crosstab_result = pd.crosstab(A['Department'], A['Gender'])
print("\nCrosstab: Employees by Department and Gender")
print(crosstab_result)



```
