## Q1. Create the DataFrame

Create a DataFrame with:

| Name | Age | Score | City | Height |
|---|---:|---:|---|---:|
| Aman | 20 | 85 | Delhi | 5.7 |
| Neha | 21 | 92 | Mumbai | 5.4 |
| Ravi | 19 | 78 | Pune | 5.8 |
| Priya | 22 | 88 | Indore | 5.5 |
| Karan | 20 | 95 | Bhopal | 5.9 |
| Simran | 21 | 90 | Delhi | 5.6 |

**Expected Output:** 6 rows × 5 columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ['Aman', 'Neha', 'Ravi', 'Priya', 'Karan', 'Simran'],
    "Age": [20, 21, 19, 22, 20, 21],
    "Score": [85, 92, 78, 88, 95, 90],
    "City": ['Delhi', 'Mumbai', 'Pune', 'Indore', 'Bhopal', 'Delhi'],
    "Height": [5.7, 5.4, 5.8, 5.5, 5.9, 5.6]
})
print(df)
```

## Q14. Create missing values

Create a copy and make Neha's Age and Ravi's Marks `np.nan`. **Expected Output:** One missing Age and one missing Marks.

```python
import numpy as np
df_copy = df.copy()
df_copy.loc[1, 'Age'] = np.nan
df_copy.loc[2, 'Score'] = np.nan
print(df_copy)
```

## Q15. Find missing values

Use `isnull().sum()`. **Expected Output:** Age = 1, Marks = 1, others = 0.

```python
df_copy.isnull().sum()
```

## Q16. Fill missing Age

Fill missing Age with 21 using `fillna()`. **Expected Output:** Neha's Age becomes 21.

```python
df_copy['Age'] = df_copy['Age'].fillna(21)
print(df_copy)
```

## Q17. Fill missing Marks

Fill missing Marks with the Marks-column mean. **Expected Output:** Ravi's missing Marks is replaced by the mean.

```python
df_copy['Score'] = df_copy['Score'].fillna(df_copy['Score'].mean())

print(df_copy)
```
