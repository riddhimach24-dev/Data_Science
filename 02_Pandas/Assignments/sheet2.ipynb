````python
a = {
    "Name": ['Aman', 'Neha', 'Ravi', 'Priya', 'Karan'],
    "Age": [20, None, 19, 22, 20],
    "Marks": [85, 92, None, 88, 95],
    "City": ['Delhi', 'Mumbai', 'Pune', 'Indore', None]

}
df = pd.DataFrame(a)
print(df)
````

## Q8. Fill missing values in the Age column with **21** using `fillna()`.

````python
df['Age'] = df['Age'].fillna(21)
print(df)
````

## Q9. Fill missing values in the Marks column with the column mean.

````python
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print(df)
````

## Q10. Fill missing values in the City column with **Unknown**.

````python
df['City'] = df['City'].fillna('Unknown')
print(df)
````
