# 🐼 Pandas Mixed Assignment (Class 1-4)

Complete all the questions using Pandas.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman","Neha","Ravi","Priya","Karan","Neha","Rohit","Simran"],
    "Age": [20,21,np.nan,22,20,21,23,22],
    "City": ["Delhi","Mumbai","Pune","Indore","Delhi","Mumbai","Pune","Delhi"],
    "Marks": [85,92,78,np.nan,95,92,88,90],
    "Course": ["Python","Data Analyst","Python","AI","Data Analyst","Data Analyst","AI","Python"]
}

df = pd.DataFrame(data)
df
```

### Q1. Display the first 5 rows of the DataFrame.

```python
df.head(5)
```

### Q2. Display the last 3 rows of the DataFrame.

```python
df.tail(3)
```

### Q3. Print all column names.

```python
df.columns
```

### Q4. Print the shape of the DataFrame.

```python
df.shape
```

### Q5. Display only the Name and Marks columns.

```python
df[['Name', 'Marks']]

```

### Q6. Display only the Age column.

```python
df['Age']
```

### Q7. Display students whose Marks are greater than 90.

```python
df[df['Marks'] > 90]
```

### Q8. Display students whose City is Delhi.

```python
df[df['City'] == 'Delhi']
```

### Q9. Display students whose Age is greater than 20.

```python
df[df['Age'] > 20]
```

### Q10. Display students whose Course is Python.

```python
df[df['Course'] == 'Python']
```

### Q11. Display students whose Course is Data Analyst and City is Mumbai.

```python
df[(df['Course'] == 'DataAnalyst') | (df['City'] == 'Mumbai')]
```

### Q12. Find the total number of missing values in each column.

```python
df.isna().sum()
```

### Q13. Fill missing values of Age with 20.

```python
df['Age'] = df['Age'].fillna(20)
df
```

### Q14. Fill missing values of Marks using the column mean.

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df
```

### Q15. Remove duplicate rows.

```python
df = df.drop_duplicates()
df
```

### Q16. Rename the column Marks to Score.

```python
df = df.rename(columns={'Marks':'Score'})
df
```

### Q17. Display students whose names start with 'N'.

```python
df[df['Name'].str.startswith('N')]
```

### Q18. Display students whose City is either Delhi or Pune.

```python
df[(df['City'] == 'Delhi') | (df['City'] == 'Pune')]
```

### Q19. Count how many students belong to each City.

```python
df['City'].value_counts()
```

### Q20. Display students whose Marks are between 80 and 95.

```python
df[(df['Marks'] >= 80) & (df['Marks'] <= 95)]
```
