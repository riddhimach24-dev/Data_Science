# 🐼 Pandas - Class 4: Data Selection & Filtering
Welcome to **Class 4** of our Pandas series. Today we’ll explore techniques to select and filter data in a DataFrame.

```python
## CREATE MISSING VALUES
# using .loc() and assign value as np.nan
import pandas as pd

df = pd.DataFrame({
    "Name": ['Aman', 'Neha', 'Ravi', 'Priya', 'Karan', 'Simran'],
    "Age": [20, 21, 19, 22, 20, 21],
    "Score": [85, 92, 78, 88, 95, 90],
    "City": ['Delhi', 'Mumbai', 'Pune', 'Indore', 'Bhopal', 'Delhi'],
    "Height": [5.7, 5.4, 5.8, 5.5, 5.9, 5.6]
})
print(df)


#Creating a copy of a dataframe
df1 = df.copy()
df1

#making Neha's age and Ravi's marks as null
df1.loc[1, 'Age'] = None
df1.loc[2, 'Score'] = np.nan
print(df1)
```

## 1. Boolean Indexing & Conditional Selection
- Use comparison operators to build conditions (e.g., `df['col'] > 50`).
- Pass the condition inside `df[...]` to get matching rows.
- Combine multiple conditions with `&` (AND), `|` (OR), and `~` (NOT).
- Parentheses are required when combining conditions.

```python
import pandas as pd


a = pd.DataFrame({
    "Name" : ['Tanishq Patel', 'Atishay', 'Krishna', 'Raj', 'Agam'],
    "Age" : [21,39,23,42,25],
    "Score" : [70,65,60,80,90]
})
print(a)

#1. Single Condition: Rows where age>30

#a['Age'] > 30    #Boolean indexing aaygi (True/False)
a[a['Age'] > 30]
a[a['Score'] >= 80]


#2. Multiple Conditions: Age>=25 AND Score>80

a[(a['Age'] >= 25) & (a['Score'] > 80)]     #AND (&)
'''(condition1) operator(and/or/not) (conndition2)'''

a[(a['Age'] >= 25) | (a['Score'] > 80)]    #OR (|)

a[(a['Age'] >= 25) | (a['Score'] > 80)]    #NOT (~)

a[~(a['Score'] > 80)]
```

## 2. Filtering Rows with Multiple Conditions
- Combine two or more conditions to refine selection.
- Remember to enclose each condition in parentheses.

```python
#1. Age > 25 AND Score >= 85
#2. Age < 35 OR Score >= 80
#3. Age b/w 25 and 35 (inclusive.)

'''Solution 3'''
a[(a['Age'] >= 25) & (a['Age'] <=35)]
```

## 3. Using `query()` for Cleaner Filtering
- `query()` allows filtering rows with a string expression.
- It's concise and readable for complex conditions.
- Variables can be passed using the `@` symbol.

```python
#Simple query: Age>25
a.query("Age > 25")     #query mai boolean wali cheez nhi hoti

#Age < 35 and Score >= 85    #Query ke andr prompt simple english mai dena hai
a.query("Age<35 and Score>=85")

s = 80
a.query("Score > @s")    #@ ke baad variable aayga
```

## 4. String Operations on Columns
- Use the `.str` accessor for vectorized string methods.
- Common methods: `contains`, `replace`, `startswith`, `endswith`, `lower`, `upper`.
- Useful for cleaning or filtering text data.

```python
#1. Convert to LowerCase
a['Name'].str.lower()

#2. Convert to UpperCase
a['Name'].str.upper()

#3. Check if name contains the letter 'a'
a[a['Name'].str.contains('a')]

#4. Replace a substring to something else
a['Name'].str.replace('Agam','Agam Jain')

#5. Filter rows where name starts with 'D'
a[a['Name'].str.startswith('a')]

#6. Filter rows where name ends with 'e'
a[a['Name'].str.endswith('a')]
```

## Mini Practice
1. Build a DataFrame with columns such as Name, Age, City, Score.
2. Use boolean indexing to select rows where Age > 25.
3. Filter rows where City is either 'Delhi' or 'Mumbai'.
4. Use query() to find rows with Score >= 85 and Age < 35.
5. Clean a text column using str.replace or str.lower.

```python


# 1. Build a DataFrame with columns: Name, Age, City, Score
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman","Neha","Ravi","Priya","Karan","Neha","Rohit","Simran"],
    "Age": [20,21,26,22,20,31,23,34],
    "City": ["Delhi","Mumbai","Pune","Indore","Delhi","Mumbai","Pune","Delhi"],
    "Score": [85,92,78,np.nan,95,92,88,90],
}

df = pd.DataFrame(data)
df

# 2. Use boolean indexing to select rows where Age > 25
df[df['Age'] > 25]

# 3. Filter rows where City is either 'Delhi' or 'Mumbai'
df[(df['City'] == 'Delhi') | (df['City'] == 'Mumbai')]


# 4. Use query() to find rows with Score >= 85 and Age < 35
df.query(" Score >= 85 and Age < 35")

# 5. Clean a text column using str.replace or str.lower
# Example: convert City names to lowercase or replace "delhi" with "New Delhi"
df['City'].str.lower()
df['City'].str.replace('Delhi', 'NewDelhi')
df
```
